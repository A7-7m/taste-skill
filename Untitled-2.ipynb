```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
print("\nDescriptive Statistics for final_score:")
print(df['final_score'].describe())
```

```python
dept_avg = df.groupby('department')['final_score'].mean().reset_index()
dept_avg = dept_avg.sort_values(by='final_score', ascending=False)

plt.figure(figsize=(8, 5))
sns.barplot(data=dept_avg, x='department', y='final_score', palette='Blues_r')
plt.title('Average Final Score by Academic Department', fontsize=12, fontweight='bold')
plt.xlabel('Department', fontsize=10)
plt.ylabel('Average Final Score', fontsize=10)
plt.ylim(0, 100)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig('task2_categorical_comparison.png')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['study_hours_per_week'].dropna(), kde=True, color='teal', bins=15)
plt.title('Distribution of Weekly Study Hours', fontsize=12, fontweight='bold')
plt.xlabel('Study Hours Per Week', fontsize=10)
plt.ylabel('Frequency', fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig('task3_distribution.png')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='attendance_pct', y='final_score', alpha=0.7, color='purple')
sns.regplot(data=df, x='attendance_pct', y='final_score', scatter=False, color='orange')
plt.title('Relationship Between Attendance Percentage and Final Score', fontsize=12, fontweight='bold')
plt.xlabel('Attendance Percentage (%)', fontsize=10)
plt.ylabel('Final Score', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig('task4_relationship.png')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='year_group', y='project_score', palette='Set2')
plt.title('Improved Visualization: Project Scores by Year Group', fontsize=12, fontweight='bold')
plt.xlabel('Year Group', fontsize=10)
plt.ylabel('Project Score', fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig('task5_redesign.png')
plt.show()
```
